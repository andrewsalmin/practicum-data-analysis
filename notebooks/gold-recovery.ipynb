### Проверка расчёта эффективности обогащения

Проверим, что эффективность обогащения рассчитана правильно. Вычислим её на обучающей выборке для признака rougher.output.recovery:

```python
C = train_data['rougher.output.concentrate_au']
F = train_data['rougher.input.feed_au']
T = train_data['rougher.output.tail_au']

rougher_calculation_recovery = C * (F - T) / (F * (C - T)) * 100
```

Найдём MAE между расчётом и значением признака:

```python
mean_absolute_error(train_data['rougher.output.recovery'], rougher_calculation_recovery)
```

Крайне малое значение MAE говорит о том, что эффективность обогащения рассчитана правильно.

### Функция для вычисления итоговой sMAPE

Напишем функцию для вычисления sMAPE:

```python
def calculate_smape(target, prediction):
    target = np.asarray(target, dtype=float)
    prediction = np.asarray(prediction, dtype=float)

    numerator = np.abs(target - prediction)
    denominator = (np.abs(target) + np.abs(prediction)) / 2

    smape = np.divide(numerator, denominator, out=np.zeros_like(numerator), where=denominator != 0)
    return np.mean(smape) * 100
```

Если и цель, и прогноз в какой-то точке равны нулю, знаменатель формулы обращается в ноль — в этом случае будем считать ошибку в этой точке нулевой (идеальное предсказание), а не `NaN`.

Напишем функцию для вычисления итоговой sMAPE:

```python
def calculate_final_smape(target, prediction):
    rougher_output_recovery_target = target['rougher.output.recovery']
    final_output_recovery_target = target['final.output.recovery']
    rougher_output_recovery_prediction = prediction[:, 0]
    final_output_recovery_prediction = prediction[:, 1]

    rougher_output_recovery_smape = calculate_smape(rougher_output_recovery_target, rougher_output_recovery_prediction)
    final_output_recovery_smape = calculate_smape(final_output_recovery_target, final_output_recovery_prediction)

    return 0.25 * rougher_output_recovery_smape + 0.75 * final_output_recovery_smape
```

### Обучение моделей

Для обучения моделей необходимо дополнить test_data недостающими столбцами из full_data. Это можно сделать, взяв ряды из full_data на основе индекса:

```python
test_data = full_data.loc[full_data.index.isin(test_data.index)]
```

Проверим, что в test_data не появились пропуски в значениях:

```python
test_data.isna().sum().sort_values(ascending=False)
```

Пропущенных значений в test_data нет.

Валидационная выборка в исходных данных отсутствует. Вместо того чтобы отделять её вручную одним случайным разбиением, будем подбирать гиперпараметры моделей с помощью кросс-валидации: она использует всю обучающую выборку и даёт более надёжную оценку качества, чем один статичный валидационный набор.

Подготовим наборы данных для обучения и проверки моделей:

```python
train_features = train_data.drop(['rougher.output.recovery', 'final.output.recovery'], axis=1)
train_target = train_data[['rougher.output.recovery', 'final.output.recovery']]

test_features = test_data.drop(
    [
        'rougher.output.recovery', 'final.output.recovery',
        'rougher.input.feed_sum', 'rougher.output.concentrate_sum',
        'primary_cleaner.output.concentrate_sum', 'final.output.concentrate_sum',
    ],
    axis=1,
)
test_target = test_data[['rougher.output.recovery', 'final.output.recovery']]
```

Проверим размерность наборов данных:

```python
print(train_features.shape)
print(train_target.shape)
print(test_features.shape)
print(test_target.shape)
```

Размерности наборов данных корректны. Можно обучать модели.

Зададим общий для моделей гиперпараметр random_state:

```python
random_state = 12345
```

Для сравнения моделей на одинаковых разбиениях зададим общую схему кросс-валидации (5 фолдов со случайным перемешиванием) и метрику — итоговую sMAPE, обёрнутую в `make_scorer` (меньшее значение лучше, поэтому `greater_is_better=False`):

```python
final_smape_scorer = make_scorer(calculate_final_smape, greater_is_better=False)
cv = KFold(n_splits=5, shuffle=True, random_state=random_state)
```

### Линейная регрессия (базовая модель для сравнения)
Прежде чем перебирать гиперпараметры деревьев, оценим на тех же фолдах простую линейную модель без гиперпараметров — это ориентир, с которым удобно сравнивать более сложные модели:

```python
linear_model = LinearRegression()
linear_cv_scores = cross_val_score(
    linear_model, train_features, train_target,
    scoring=final_smape_scorer, cv=cv, n_jobs=-1,
)

linear_best_smape = -linear_cv_scores.mean()
linear_best_smape_std = linear_cv_scores.std()
print(f'sMAPE (CV) = {linear_best_smape:.4f} ± {linear_best_smape_std:.4f}')
```

Зафиксируем модель на всей обучающей выборке — она понадобится для итоговой таблицы сравнения моделей:

```python
linear_model.fit(train_features, train_target)
```

Подберём гиперпараметр `max_depth` модели DecisionTreeRegressor с помощью кросс-валидации (GridSearchCV). Возьмём диапазон шире, чем скорее всего потребуется, и добавим `None` (нет ограничения глубины) — тогда GridSearchCV сам покажет, где начинается переобучение, а не упрётся в границу перебора:

```python
dtr_param_grid = {'max_depth': [*range(1, 13), None]}

dtr_grid_search = GridSearchCV(
    DecisionTreeRegressor(random_state=random_state),
    param_grid=dtr_param_grid,
    scoring=final_smape_scorer,
    cv=cv,
    n_jobs=-1,
)
dtr_grid_search.fit(train_features, train_target)

best_model_dtr = dtr_grid_search.best_estimator_
dtr_best_smape = -dtr_grid_search.best_score_
dtr_best_smape_std = dtr_grid_search.cv_results_['std_test_score'][dtr_grid_search.best_index_]
dtr_best_max_depth = dtr_grid_search.best_params_['max_depth']
```

Наименьшее значение sMAPE по результатам кросс-валидации:

```python
print(f'sMAPE (CV) = {dtr_best_smape:.4f} ± {dtr_best_smape_std:.4f}')
```

Наименьшее значение sMAPE получено при следующих значениях гиперпараметров:

```python
print('max_depth =', dtr_best_max_depth)
print('random_state =', random_state)
```

Подберём гиперпараметры `n_estimators` и `max_depth` модели RandomForestRegressor с помощью кросс-валидации (GridSearchCV), также на расширенной сетке значений (включая `max_depth=None`). Переберём все комбинации на тех же фолдах, что и для предыдущих моделей; включим `verbose`, чтобы отслеживать прогресс — перебор занимает время:

```python
rfr_param_grid = {
    'n_estimators': range(10, 81, 10),
    'max_depth': [*range(1, 13), None],
}

rfr_grid_search = GridSearchCV(
    RandomForestRegressor(random_state=random_state),
    param_grid=rfr_param_grid,
    scoring=final_smape_scorer,
    cv=cv,
    n_jobs=-1,
    verbose=1,
)
rfr_grid_search.fit(train_features, train_target)

best_model_rfr = rfr_grid_search.best_estimator_
rfr_best_smape = -rfr_grid_search.best_score_
rfr_best_smape_std = rfr_grid_search.cv_results_['std_test_score'][rfr_grid_search.best_index_]
rfr_best_n_estimators = rfr_grid_search.best_params_['n_estimators']
rfr_best_max_depth = rfr_grid_search.best_params_['max_depth']
```

Наименьшее значение sMAPE по результатам кросс-валидации:

```python
print(f'sMAPE (CV) = {rfr_best_smape:.4f} ± {rfr_best_smape_std:.4f}')
```

Наименьшее значение sMAPE получено при следующих значениях гиперпараметров:

```python
print('n_estimators =', rfr_best_n_estimators)
print('max_depth =', rfr_best_max_depth)
print('random_state =', random_state)
```

Модель RandomForestRegressor показала лучший показатель sMAPE. Возьмём данную модель за основную.
